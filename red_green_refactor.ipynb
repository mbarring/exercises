### Following the **Test-Driven Development (TDD)** approach as outlined in the slides, here's a **simple Python code example** implementing the **Red-Green-Refactor** cycle.

### **What is the Core Task of This Example?**
The **core task** of this example is to **implement the `check_threshold()` function using Test-Driven Development (TDD) and verify its behavior with a structured test**.

---

### **🔍 Task Breakdown**
#### **1️⃣ Objective**
Implement a function `check_threshold(number, minimum_threshold)` that:
- **Returns `True` if `number >= minimum_threshold`**.
- **Returns `False` otherwise**.

#### **2️⃣ Methodology**
- **Follow the Test-Driven Development (TDD) approach** using the **Red-Green-Refactor** cycle:
  - **RED (Failing Test)**: Write a **test first**, which will fail because `check_threshold()` is not yet implemented.
  - **GREEN (Minimal Implementation)**: Implement the **simplest code** to pass the test (even if it’s a hardcoded `return True`).
  - **REFACTOR (Optimize the Code)**: Improve the function’s logic to match the real requirement, ensuring the test still passes.

- **Key Test Principles**
  - **Test only one requirement** (`number >= minimum_threshold`).
  - Use `assert` statements to validate expected behavior.
  - **Handle exceptions** to prevent test crashes.
  - **Decouple the test from the function implementation** by passing the function as a parameter.

---

## **1️⃣ Step 1: RED – Write a Failing Test**
- **Before implementing `check_threshold()`**, we first write a test.
- This test **will fail** because `check_threshold()` is not yet defined.


### **🚨 Expected Output**
```

```
This failure is **intentional** because `check_threshold()` is not yet implemented.

## **2️⃣ Step 2: GREEN – Implement the Simplest Code to Pass the Test**
- Now, we **define `check_threshold()` with a hardcoded `True` return value** to make the test pass.

### **✅ Expected Output**
```
✅ Test Passed
```
Although the test passes, this implementation **isn't correct for all cases**, so we need to refactor.



````python
def check_threshold(number, minimum_threshold):
    """
    Simplest implementation to pass the test (initial step).
    """
    return True  # ✅ Hardcoded return to make the test pass


def test_check_threshold(function_under_test):
    """
    Tests check_threshold() for the requirement:
    - Returns True when number >= minimum_threshold.
    - Handles exceptions.
    - Returns True (Pass) or False (Fail).
    """
    try:
        # Test case: 8 >= 4, expected output: True
        assert function_under_test(8, 4) == True
        print("✅ Test Passed")
        return True  # Pass
    except AssertionError:
        print("❌ Test Failed: Expected True but got False.")
        return False  # Fail
    except Exception as e:
        print(f"❌ Test Failed due to an Exception: {e}")
        return False  # Fail

# Run the test
passed = test_check_threshold(check_threshold)


````

## **3️⃣ Step 3: REFACTOR – Implement the Correct Logic**
- Now, we replace the hardcoded return value with the correct comparison logic.


### **✅ Expected Output**
```
✅ Test Passed
```
This time, the test passes with a correct implementation.



````python
def check_threshold(number, minimum_threshold):
    """
    Compares number with minimum_threshold.
    Returns True if number >= minimum_threshold, otherwise returns False.
    """
    return number >= minimum_threshold  # ✅ Correct implementation


def test_check_threshold(function_under_test):
    """
    Tests check_threshold() for the requirement:
    - Returns True when number >= minimum_threshold.
    - Handles exceptions.
    - Returns True (Pass) or False (Fail).
    """
    try:
        # Test case: 8 >= 4, expected output: True
        assert function_under_test(8, 4) == True
        print("✅ Test Passed")
        return True  # Pass
    except AssertionError:
        print("❌ Test Failed: Expected True but got False.")
        return False  # Fail
    except Exception as e:
        print(f"❌ Test Failed due to an Exception: {e}")
        return False  # Fail

# Run the test
passed = test_check_threshold(check_threshold)

````


## **4️⃣ (Optional) Add More Test Cases**
- Now, let's expand our tests to **cover edge cases**.


### **✅ Expected Output**
```
Test Case #1: check_threshold(8, 4)
✅ Test Passed
Test Case #2: check_threshold(4, 4)
✅ Test Passed
Test Case #3: check_threshold(3, 4)
✅ Test Passed
Test Case #4: check_threshold(-1, 0)
✅ Test Passed
Test Case #5: check_threshold(100, 50)
✅ Test Passed
```



````python
def check_threshold(number, minimum_threshold):
    """
    Compares number with minimum_threshold.
    Returns True if number >= minimum_threshold, otherwise returns False.
    """
    return number >= minimum_threshold  # Correct implementation


def test_check_threshold(function_under_test):
    """
    Tests the check_threshold function with multiple test cases.
    Uses assert statements to validate expected behavior.
    Handles exceptions and prints test results.
    """
    test_cases = [
        (8, 4, True),   # Normal case: 8 >= 4
        (4, 4, True),   # Edge case: 4 == 4
        (3, 4, False),  # Failing case: 3 < 4
        (-1, 0, False), # Negative numbers case
        (100, 50, True) # Large numbers case
    ]

    for i, (num, threshold, expected) in enumerate(test_cases, start=1):
        print(f"Test Case #{i}: check_threshold({num}, {threshold})")
        try:
            assert function_under_test(num, threshold) == expected
            print("✅ Test Passed")
        except AssertionError:
            print(f"❌ Test Failed: Expected {expected} but got {not expected}")
        except Exception as e:
            print(f"❌ Test Failed due to an Exception: {e}")

# Run all test cases
test_check_threshold(check_threshold)
````
